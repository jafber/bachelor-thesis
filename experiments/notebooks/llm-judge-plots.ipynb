# LLM judge: exact first-error evaluation

Full-run Qwen3.5 CSVs (`runs/llm_judge_qwen35_*_full_arm.csv`) and cached Llama 3.1/3.2 responses (`cache/llm_judge_llama*.sqlite`), no inference. Exclude the broken BIG-Bench Mistake Dyck trace with UID 52 from **all models**. All reported percentages use the remaining trace denominator: malformed/out-of-range answers are **wrong**, including on clean traces. The binary MR-GSM8K MCC counts invalid responses as incorrect binary predictions; a valid but wrong error *location* still counts as identifying an erroneous trace. The published numbers are baselines from different models/prompts and are not replications. All paper scores below are percentages (MCC multiplied by 100).

```python
from pathlib import Path
import sys

EXPERIMENTS = Path.cwd().parent
sys.path.insert(0, str(EXPERIMENTS))
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from IPython.display import display
from experiment_setup.benchmarks import load_examples
from experiment_setup import llm_critic
from experiment_setup.judge_metrics import summarize, processbench_average

RUNS = EXPERIMENTS / 'runs'
CACHE = EXPERIMENTS / 'cache'
OUT = EXPERIMENTS.parent / 'thesis/figures'
MODELS = {
    'Qwen3.5-4B': ('Qwen/Qwen3.5-4B', 'qwen35_4b'),
    'Qwen3.5-9B': ('Qwen/Qwen3.5-9B', 'qwen35_9b'),
    'Qwen3.5-27B': ('Qwen/Qwen3.5-27B', 'qwen35_27b'),
    'Llama3.1-8B': ('meta-llama/Llama-3.1-8B-Instruct', 'llama31_8b'),
    'Llama3.2-3B': ('meta-llama/Llama-3.2-3B-Instruct', 'llama32_3b'),
}
BENCHMARKS = ['MR-GSM8K', 'BIG-Bench-Mistake', 'ProcessBench']
examples = [e for task in load_examples().values() for e in task
            if not (e.benchmark == 'BIG-Bench-Mistake' and e.task == 'dyck_languages' and e.uid == '52')]
results = {}
for model, (model_id, stem) in MODELS.items():
    path = CACHE / f'llm_judge_{stem}_arm.sqlite'
    frame = llm_critic.load_run(examples, model_id=model_id, cache_path=path)
    assert frame.groupby('benchmark').size().to_dict() == dict(zip(BENCHMARKS, [2999, 2185, 3400])), path
    assert not frame.duplicated(['benchmark', 'task', 'uid']).any(), path
    results[model] = frame
```

```python
def paper_comparison(ax, paper, ours, source, metric, max_y=100):
    """Ascending score order, with labeled bars and one legend per chart."""
    values = pd.Series(paper | ours).sort_values(kind='stable')
    colors = ['tab:gray' if name in paper else 'tab:blue' for name in values.index]
    values.plot.barh(ax=ax, color=colors)
    left = min(0, values.min())
    ax.set(xlabel=metric, ylabel='',
           xlim=(left - max_y * 0.05 if left < 0 else 0, max_y * 1.12))
    if left < 0:
        ax.axvline(0, linestyle='--', color='tab:gray')
    label_format = '%.3f' if max_y == 1 else '%.1f'
    bars = ax.containers[0]
    ax.bar_label(bars, labels=[label_format % value if value >= 0 else '' for value in values],
                 padding=3, fontsize=8)
    # Negative bar labels placed to the left collide with long model names;
    # print the signed value just to the right of the zero line instead.
    for bar, value in zip(bars, values):
        if value < 0:
            ax.annotate(label_format % value, (0, bar.get_y() + bar.get_height() / 2),
                        xytext=(5, 0), textcoords='offset points', va='center', fontsize=8)
    ax.legend(handles=[Patch(color='tab:gray', label=source),
                       Patch(color='tab:blue', label='Ours')], loc='lower right')
```

## Trace accuracy

Overall = exact first-error location (or NONE); clean = exact NONE on error-free traces; error = exact first error on erroneous traces. Invalid outputs are incorrect in each metric.

```python
overview = pd.concat([summarize(frame).assign(model=model) for model, frame in results.items()], ignore_index=True)
overview['benchmark'] = pd.Categorical(overview.benchmark, BENCHMARKS, ordered=True)
overview['model'] = pd.Categorical(overview.model, list(MODELS), ordered=True)
overview = overview.sort_values(['benchmark', 'model'])
display(overview[['benchmark', 'model', 'n', 'errors', 'unparseable', 'accuracy', 'correct_accuracy', 'error_accuracy']].style.format({
    'accuracy':'{:.1%}', 'correct_accuracy':'{:.1%}', 'error_accuracy':'{:.1%}'}))
```

```python
ax = overview.pivot(index='benchmark', columns='model', values='accuracy').reindex(index=BENCHMARKS, columns=MODELS).mul(100).plot.bar(figsize=(12, 5), rot=0)
ax.set(xlabel='', ylabel='Exact first-error accuracy (%)')
ax.legend(title='Model')
ax.set_ylim(top=100)
plt.tight_layout()
plt.show()
```

```python
ax = overview.pivot(index='benchmark', columns='model', values='error_accuracy').reindex(index=BENCHMARKS, columns=MODELS).mul(100).plot.bar(figsize=(12, 5), rot=0)
ax.set(xlabel='', ylabel='Exact first-error accuracy (%)')
ax.legend(title='Model')
ax.set_ylim(top=100)
plt.tight_layout()
plt.show()
```

## ProcessBench — Zheng et al. (2025), Table 3 (p. 7)

All **open-weight language models prompted as critics** from the paper (not PRMs or proprietary models) are shown. The paper calls the harmonic mean of *separate* clean/error exact-location accuracies “F1”. Its **Average** is the unweighted mean of the four split harmonic means, **not** pooled F1 or pooled harmonic mean. Only this benchmark-level average is plotted. Paper open-weight critic results use majority vote over eight samples (our runs use one response); prompts and model generations also differ.

## BIG-Bench Mistake — paper Table 4 (p. 4), direct trace

Compare **exact first-error accuracy over all included traces**, not weighted binary F1 (paper Table 5 predicts *final-answer correctness*, which is not the same as process correctness). The paper's “Overall” is the unweighted mean of the five task accuracies. We calculate the same macro average, but our Dyck task uses **985 traces** (excluding broken UID 52) rather than the paper's 592-trace subsample (GPT-4-Turbo: 300), so the overall figures are **not** directly comparable. Paper uses three-shot, task-specific direct-trace prompts; our judge is zero-shot with a shared prompt.

```python
# Tyen et al. (2024), Table 4, Overall / Direct (trace) column: all five models.
bbm_paper = {
    'GPT-4-Turbo': 30.13,
    'GPT-4': 39.80,
    'GPT-3.5-Turbo': 10.44,
    'Gemini Pro': 16.14,
    'PaLM 2 Unicorn': 17.09,
}
assert len(bbm_paper) == 5
bbm_ours = {model: summarize(
    frame[frame.benchmark == 'BIG-Bench-Mistake'], ('task',)
).accuracy.mean() * 100 for model, frame in results.items()}
fig, ax = plt.subplots(figsize=(10, 6))
paper_comparison(ax, bbm_paper, bbm_ours, 'Tyen et al. (2024)',
                 'Macro-average exact first-error accuracy (%)')
fig.tight_layout()
fig.savefig(OUT / 'tyen.png', dpi=220, bbox_inches='tight')
plt.show()
```

```python
# Thesis figure: conditional exact accuracy separates localization from saying NONE.
out = OUT
fig, axes = plt.subplots(1, 3, figsize=(17, 4), sharey=True, constrained_layout=True)
for ax, benchmark in zip(axes, BENCHMARKS):
    subset = overview[overview.benchmark == benchmark].set_index('model').reindex(MODELS)
    subset[['correct_accuracy', 'error_accuracy']].rename(columns={
        'correct_accuracy': 'Clean traces: NONE', 'error_accuracy': 'Error traces: first error'
    }).mul(100).plot.bar(ax=ax, rot=35, legend=False)
    ax.set(title=benchmark, xlabel='', ylabel='Exact accuracy (%)', ylim=(0, 100))
    ax.grid(axis='y', alpha=.2)
fig.legend(axes[0].get_legend_handles_labels()[0], ['Clean traces: NONE', 'Error traces: first error'],
           loc='lower center', ncol=2, bbox_to_anchor=(.5, -.12))
fig.savefig(out / 'judge-conditional-accuracy.png', dpi=220, bbox_inches='tight')
plt.show()
```
