# Thinking-mode decoding: greedy vs. recommended sampling

```python
import gc

import torch

from llm_critic import TransformersJudge, load_examples, prompt_for
```

```python
examples = load_examples()
sample = examples["MR-GSM8K", "original"][0]
```

```python
prompt = prompt_for(sample)
model_id = "Qwen/Qwen3.5-4B"

judge = TransformersJudge(
    model_id, thinking=True, max_new_tokens=4096,
    generation_kwargs={"do_sample": False, "top_p": 1.0, "top_k": 50,
                       "min_p": None, "presence_penalty": 0.0},
)
print("=== GREEDY: COMPLETE RAW PROMPT + RESPONSE (including special tokens) ===")
print(judge.generate_batch([prompt])[0])
del judge
gc.collect()
torch.cuda.empty_cache()

judge = TransformersJudge(model_id, thinking=True, max_new_tokens=4096)
print("=== DEFAULT SAMPLING: COMPLETE RAW PROMPT + RESPONSE (including special tokens) ===")
print(judge.generate_batch([prompt])[0])
```
